```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import json
import pandas as pd

# Load JSON (can also load from file using json.load())
data = [
  {
    "order_id": 1,
    "customer": {
      "name": "Alice",
      "region": "East"
    },
    "items": [
      {"product": "Pen", "quantity": 2, "price": 1.5},
      {"product": "Notebook", "quantity": 1, "price": 3.0}
    ]
  },
  {
    "order_id": 2,
    "customer": {
      "name": "Bob",
      "region": "West"
    },
    "items": [
      {"product": "Pencil", "quantity": 3, "price": 0.5}
    ]
  }
]

# Normalize JSON: one row per item
df = pd.json_normalize(data, record_path="items", meta=["order_id", ["customer", "name"], ["customer", "region"]])
print(df)
```

```
import seaborn as sns
import matplotlib.pyplot as plt

product_qty = df.groupby("product")["quantity"].sum().reset_index()

sns.barplot(x="product", y="quantity", data=product_qty)
plt.title("Total Quantity Sold per Product")
plt.show()

Example 2: Region-wise sales amount (quantity × price)

df["sales"] = df["quantity"] * df["price"]
region_sales = df.groupby("customer.region")["sales"].sum().reset_index()

sns.barplot(x="customer.region", y="sales", data=region_sales)
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```
import pandas as pd
import numpy as np

# Create a sample sales dataset
np.random.seed(42)
months = pd.date_range(start='2024-01-01', periods=12, freq='M')
teams = ['Team A', 'Team B', 'Team C']
data = []

for team in teams:
    sales = np.random.randint(20000, 50000, size=len(months)).cumsum()  # cumulative sales
    for month, sale in zip(months, sales):
        data.append({'Month': month, 'Team': team, 'Cumulative_Sales': sale})

df = pd.DataFrame(data)
df.head()
```

```
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='Month', y='Cumulative_Sales', hue='Team', marker='o')
plt.title('Sales Growth Over Time by Team')
plt.ylabel('Cumulative Sales ($)')
plt.xlabel('Month')
plt.grid(True)
plt.show()
```

```
df['Monthly_Sales'] = df.groupby('Team')['Cumulative_Sales'].diff().fillna(df['Cumulative_Sales'])

plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Month', y='Monthly_Sales', hue='Team')
plt.title('Monthly Sales by Team')
plt.ylabel('Monthly Sales ($)')
plt.xlabel('Month')
plt.xticks(rotation=45)
plt.show()

```

```
import plotly.express as px

fig = px.line(df, x='Month', y='Cumulative_Sales', color='Team', markers=True,
              title='Interactive Sales Growth Over Time')
fig.show()
```
